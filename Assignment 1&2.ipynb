```python
# Write a program which takes 5 inputs from user for different subject's marks, total it and generate mark sheet using grades?


print(" Mark Sheet ")

math = float(input("Enter marks of math: "))
comp = float(input("Enter marks of comp: "))
urdu = float(input("Enter marks of urdu: "))
isla = float(input("Enter marks of isla: "))
sci = float(input("Enter marks of sci: "))

total = math + comp + urdu + isla + sci
percentage = (total / 500) * 100

if percentage >= 80:
    grade = "A+"
elif percentage >= 70:
    grade = "A"
elif percentage >= 60:
    grade = "B"
elif percentage >= 50:
    grade = "C"
else:
    grade = "Fail"

print("\n Result ")
print("Total Marks:", total)
print("Percentage:", percentage, "%")
print("Grade:", grade)
```
